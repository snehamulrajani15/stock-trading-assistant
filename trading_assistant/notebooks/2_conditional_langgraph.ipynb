```python
from typing import TypedDict, Literal

class PortfolioState(TypedDict):
    amount_gbp: float
    total_gbp: float
    target_currency: Literal["inr", "eur"]
    total: float
```

```python
def calc_total(state: PortfolioState) -> PortfolioState:
    #8.08 is the interest earned. Will be called using api for live applications
    state['total_gbp'] = state['amount_gbp'] * 8.08
    return state

def convert_to_inr(state: PortfolioState) -> PortfolioState:
    #127 is the conversion rate. Will be called using api for live applications
    state['total_inr'] = state['total_gbp'] * 127
    return state

def convert_to_eur(state: PortfolioState) -> PortfolioState:
    #1.1 is the conversion rate
    state['total_eur'] = state['total_gbp'] * 1.1
    return state
```
